```python
import pandas as pd

df = pd.read_csv('final_df.csv')
```

```python
print(df["race"].unique().tolist())
```

```python
print(df["race"].value_counts())
```

```python
groups_to_merge = [
    "American Indian or Alaska Native",
    "Prefer not to say",
    "Native Hawaiian or other Pacific Islander",
    "Middle Eastern",
    "Not specified"
]

df["race"] = df["race"].replace(dict.fromkeys(groups_to_merge, "Other race, ethnicity or origin"))
```

```python
from sklearn.model_selection import train_test_split, StratifiedKFold

# Create a combined stratification label
df['strata_race'] = df['race'].astype(str) + '_' + df['study_group'].astype(str)

# Check counts per combination — StratifiedKFold/train_test_split will fail
# (or misbehave) if any combo has too few samples for the number of folds/splits
print(df['strata_race'].value_counts())
```

```python
from sklearn.model_selection import train_test_split, StratifiedKFold

# Create a combined stratification label
df['strata_sex'] = df['sex'].astype(str) + '_' + df['study_group'].astype(str)

# Check counts per combination — StratifiedKFold/train_test_split will fail
# (or misbehave) if any combo has too few samples for the number of folds/splits
print(df['strata_sex'].value_counts())
```

```python
bmi_col = 'bmi_vsorres, BMI'

df['weight_class'] = pd.cut(
    df[bmi_col],
    bins=[-float('inf'), 18.5, 25, 30, float('inf')],
    labels=['Underweight', 'Normal', 'Overweight', 'Obesity'],
    right=False
)

print(df['weight_class'].value_counts())
```

```python
from sklearn.model_selection import train_test_split, StratifiedKFold

# Create a combined stratification label
df['strata_weight_class'] = df['weight_class'].astype(str) + '_' + df['study_group'].astype(str)

# Check counts per combination — StratifiedKFold/train_test_split will fail
# (or misbehave) if any combo has too few samples for the number of folds/splits
print(df['strata_weight_class'].value_counts())
```

```python
df.to_csv("strata_df.csv", index=False)
```
